```python
import statistics
```

```python
numbers = [10, 20, 30, 40, 50]

print("Dataset:", numbers)
print("Number of elements:", len(numbers))
print("Sum:", sum(numbers))
```

```python
mean_value = statistics.mean(numbers)

print("Mean:", mean_value)
```

```python
median_value = statistics.median(numbers)

print("Median:", median_value)
```

```python
std_value = statistics.stdev(numbers)

print("Standard Deviation:", round(std_value, 2))
```

```python
print("Mean:", statistics.mean(numbers))
print("Median:", statistics.median(numbers))
print("Mode:", statistics.mode(numbers))
print("Standard Deviation:", round(statistics.stdev(numbers), 2))
print("Variance:", statistics.variance(numbers))
```
