```python
from collections import deque
graph = {
    'A' : ['B','C'],
    'B' : ['D','E'],
    'C' : ['F'],
    'D' : ['G'],
    'E' : [],
    'F' : ['H'],
    'G' : [],
    'H' : []
}

def bfs(graph,start):

    visited  = []  #stores visited nodes
    queue = deque() #create an empty queue

    visited.append(start) #visit the start node
    queue.append(start) #add start node to the queue

    while queue:

      node = queue.popleft() #remove first node=A
      print(node, end=" ")

      #visit all neighbours
      for neighbour in graph[node]:
        if neighbour not in visited:
          visited.append(neighbour)
          queue.append(neighbour)

#user input
start = input("Enter Start Node: ").upper()

print("BFS Traversal: ")
bfs(graph, start)
```

```python
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': ['G'],
    'E': [],
    'F': ['H'],
    'G': [],
    'H': []
}

def dfs(graph, start):

    visited = []   # Stores visited nodes
    stack = []     # Create an empty stack

    stack.append(start)

    while stack:

        node = stack.pop()   # Remove the top element

        if node not in visited:
            visited.append(node)
            print(node, end=" ")

            # Push neighbours in reverse order
            for neighbour in reversed(graph[node]):
                if neighbour not in visited:
                    stack.append(neighbour)

# User input
start = input("Enter Start Node: ").upper()

print("DFS Traversal:")
dfs(graph, start)
```
