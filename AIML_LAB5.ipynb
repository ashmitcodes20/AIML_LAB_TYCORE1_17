```python
df['Age'].fillna(df['Age'].mean(), inplace=True)
df['Embarked'].fillna(df['Embarked'].mode()[0], inplace=True)
df.drop('Cabin', axis=1, inplace=True)
```

```python
print(df.isnull().sum())
```

```python
result = df[df['Age'] > 30]

print(result)
```

```python
result = df[df['Fare'] > 50]

print(result)
```

```python
result = df[df['Age'] > 30][['Name', 'Age', 'Fare']]

print(result)
```

```python
result = df[(df['Sex'] == 'female') & (df['Survived'] == 1)]

print(result)
```

```python
result = df[(df['Pclass'] == 1) | (df['Pclass'] == 2)]

print(result)
```

```python
result = df[df['Pclass'].isin([1, 3])]

print(result)
```

```python
result = df[df['Name'].str.contains('Mrs', na=False)]

print(result)
```

```python
result = df[df['Age'].isnull()]

print(result)
```

```python
result = df.sort_values('Age')

print(result)
```

```python
result = df.sort_values(
    by=['Pclass', 'Fare']
)

print(result)
```

```python
result = df.sort_values(
    by=['Pclass', 'Fare'],
    ascending=[True, False]
)

print(result)
```

```python
result = df.sort_values(
    by=['Age', 'Fare'],
    ascending=[True, False]
)

print(result)
```

```python
result = df.sort_values(
    'Fare',
    ascending=False
)[['Name', 'Age', 'Fare']]

print(result)
```

```python
result = df.sort_values(
    'Fare',
    ascending=False
).head(10)

print(result)
```
