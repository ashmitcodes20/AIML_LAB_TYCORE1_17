```python
import matplotlib.pyplot as plt


students = ["Aaryan", "Rahul", "Sneha", "Priya", "Aditya", "Rohan"]
marks = [85, 72, 91, 68, 88, 76]


attendance = [90, 80, 95, 75, 88, 82]
```

```python
plt.plot(students, marks, marker='o', label="Marks")
plt.plot(students, attendance, marker='s', label="Attendance")

plt.title("Student Marks and Attendance")
plt.xlabel("Students")
plt.ylabel("Values")

plt.legend()
plt.grid(True)

plt.show()
```

```python
x = [1, 2, 3, 4, 5, 6]

plt.bar(x, marks, label="Marks")
plt.bar(x, attendance, bottom=marks, label="Attendance")

plt.title("Marks and Attendance")
plt.xlabel("Students")
plt.ylabel("Values")

plt.xticks(x, students)
plt.legend()

plt.show()
```

```python
x1 = [1, 2, 3, 4, 5, 6]
x2 = [value + 0.4 for value in x1]

plt.bar(x1, marks, width=0.4, label="Marks")
plt.bar(x2, attendance, width=0.4, label="Attendance")

plt.title("Marks vs Attendance")
plt.xlabel("Students")
plt.ylabel("Values")

plt.xticks(x1, students)
plt.legend()

plt.show()
```

```python
plt.scatter(marks, attendance)

plt.title("Marks vs Attendance")
plt.xlabel("Marks")
plt.ylabel("Attendance")

plt.grid(True)

plt.show()
```

```python
plt.hist(marks, bins=5)

plt.title("Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.hist(attendance, bins=5)

plt.title("Distribution of Attendance")
plt.xlabel("Attendance")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.hist(marks, bins=5, alpha=0.5, label="Marks")
plt.hist(attendance, bins=5, alpha=0.5, label="Attendance")

plt.title("Marks and Attendance Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")

plt.legend()

plt.show()
```

```python
plt.pie(
    marks,
    labels=students,
    autopct='%1.1f%%'
)

plt.title("Student Marks Distribution")

plt.show()
```

```python
plt.pie(
    attendance,
    labels=students,
    autopct='%1.1f%%'
)

plt.title("Student Attendance Distribution")

plt.show()
```

```python
plt.boxplot(
    [marks, attendance],
    tick_labels=["Marks", "Attendance"]
)

plt.title("Marks and Attendance Box Plot")
plt.ylabel("Values")

plt.show()
```
